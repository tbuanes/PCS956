---
title: Bayesian Inference
---

Even if information we have is insufficient to determine what class an instance belong to, we can still hope to calculate probabilities for it belonging to the various possible clases. I.e. we want to calculate:
$$P(\text{class}|\text{data},\text{background knowledge}).$$
Here 'data' is the measurement data for the instance we are trying to classify, and 'background knowlege' is any other relevant knowledge we have, including on the relation between different possible measurement data and the true class.

According to Bayes' theorem, we can calculate this as
$$P(\text{class}|\text{data},\text{background knowledge}) 
    = \frac{P(\text{class}|\text{background knowledge})P(\text{data}|\text{class},\text{background knowledge})}{P(\text{data}|\text{background knowledge})}
$$

Here $P(\text{class}|\text{background knowledge})$ is the so-called prior, i.e. the probability we would assign without having the specific information given by the measurement data for that instance. If we know the distribution of the different classes in the population we sample from, we would set the prior equal to that frequency distribution. $P(\text{data}|\text{class},\text{background knowledge})$ is the probability for observing the measurement data we have given the class we are predicting together with the background knowledge. Finally, $P(\text{data}|\text{background knowledge})$ is the probability for observing the measurment data given the background knowledge, but not restricting us to a specific class. 

If either the class and/or the data is replaced by a continuous variable, the expression above needs to be modified slightly. Let $p(\theta|x)$ be the probability density function (p.d.f.) describing the probability for observing a certain value of the parameter(s) of interest, $\theta$, given data $x$ (which usually will be multi-dimensional). The probability will still also depend on the background knowledge, but this denpendency is supressed to de-clutter the notation. Bayes' formula for this case is:
$$
    p(\theta|x) = \frac{p(x|\theta)p(\theta)}{\int p(x|\theta)p(\theta)\mathrm{d}\theta}
$$
which has the same interpretation as the version above. I.e., $p(x|\theta)$ is the probability for the observation we have made ($x$) given that the parameter(s) of interest are $\theta$, $p(\theta)$ is the probility for the parameter(s) of interest being $\theta$ prior to the extra information provided by observing $x$, and the integral in the denominator is the probability for observing $x$ irrespective of the value of $\theta$. 

If the goal is to determine the value of $\theta$ we are basically done now.^[Provied that we can acually compute the involved quantities.] Of course we can refine our determination of $\theta$ by performing more observations and repeat the procedure. However, we are often not primarily interested in determining the value of $\theta$, but in predicting the value $\tilde{x}$ of a new observation given the information we have already gathered through observations $\mathbf{x} = \{x_1,\ldots,x_n\}$. This can be calculated as
$$
    p(\tilde{x}|\mathbf{x}) = \int p(\tilde{x}|\theta)p(\theta|\mathbf{x})\mathrm{d}\theta
$$
Where $p(\theta|\mathbf{x})$ is the result of the calculation using Bayes' formula above (possibly multiple times to include several observations).
