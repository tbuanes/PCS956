title: Summary of Module B 
---

The red thread of the module has been to have a critical look at how we evaluate the outcome of any machine learning model to ensure that we can trust its output/decision. A machine learning model may look impressive with a high accuracy, but still be completely useless. Whether you are developing or using models, it is important to make sure that the model you are working with is robust and trusthworthy---not only gives impressive stats on some test data set.

## Learning outcomes

**Metrics**

* Have an overview of the most common metrics used in machine learning
* Understand the pros and cons of commonly used metrics
* Understand why it is important to choose a metric that is well suited for the problem at hand

**Calibration**

* Understand that the output of a machine learning classifier cannot in general be interpreted as probability (even though it may look like it)
* Have an overview of common approaches to calibration
* Understand that calibration may not be enough---even if the model does well on average, there may be large instance-to-instance variations.

**Uncertainty**

* Understand the value of having an uncertainty estimate on the machine learning prediction
* Understand the different sources of uncertainty (i.e. stages that uncertainty may enter the process) in data analysis using machine learning
* Have an overview of strategies for estimating uncertainty associated with machine learning predictions
